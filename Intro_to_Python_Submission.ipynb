# **FUNCTIONS**

```python
# --------- DEFINING A FUNCTION ----------
def celsius_to_fahrenheit(celsius):
  fahrenheit = (celsius * 9/5) + 32
  return fahrenheit
```

A function is a reusable block of code, like a code variable. This function called celsius_to_fahrenheit() converts a temperature from Celsius to Fahrenheit using a formula.

```python
# ---------- CALLING FUNCTIONS --------
celsius_to_fahrenheit(0)
```

Calling the function at the value 0 will convert from celsius to fahrenheit correctly.

# **IF STATEMENTS**

```python
# --------- IS IT COLD ----------
temp = 10

if temp < 0:
  print('It is cold')
else:
    print('It is not cold')
```

An if statement allows the program to make "decisions". If temp < 0 it prints 'It is cold'. Otherwise, it runs the else block.

```python
# --------- LOOPS ----------
Celsius_temps_to_convert = [-20, 0, 10, 20, 30, 40]

for temp in Celsius_temps_to_convert:
  print(celsius_to_fahrenheit(temp))
```

A for loop repeats a block of code for each item in a list. Here, each Celsius is converted to Fahrenheit and then printed.

```python
# --------- LOOP WITH CONDITIONAL ----------
for temp in Celsius_temps_to_convert:
  if celsius_to_fahrenheit(temp) < 32:
    print(f"temperature of {celsius_to_fahrenheit(temp)}F means it is cold")
  else:
    print(f"temperature of {celsius_to_fahrenheit(temp)}F means it is not cold")
```

This loop combines the function and if statements. Each temperature is converted, then checked:
Below 32°F = cold,
Otherwise = not cold
