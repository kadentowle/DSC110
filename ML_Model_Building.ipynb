# **IMPORTING LIBRARIES**

```python
# ---------- IMPORTS ----------

import pandas as pd
from sklearn.linear_model import LinearRegression
```

The libraries are the tools we need, pandas is used to create dataframes, where sklearn is how we'll build our machine learning model.

# **LINEAR REGRESSION (CONTINUOUS OUTPUT)**

```python
# ---------- CREATING I/O DATA ----------
X = pd.DataFrame({'age': [20, 30, 40, 50, 60]})
y =                      [110, 115, 120, 125, 130]
```

X is the input data (age).
Y is the output data (unknown).
Their relationship is Y = 0.5X + 100.

```python
# ---------- BUILDING A LINEAR REGRESSION MODEL ----------
lr = LinearRegression()
lr.fit(X, y)
```

These lines build a linear regression model and train it on the I/O data from above. The machine should learn the pattern of the data and be able to predict numbers it hasn't seen in the data provided.

```python
# ---------- MAKING NEW PREDICTIONS ---------
ages_to_predict = pd.DataFrame({'age': [10, 45, 80]})
predictions = lr.predict(ages_to_predict)

print(predictions)
```

This is where we give the model numbers it's never seen before. It has learned the linear relationship between x and y and can now create predictions.

```python
for p in predictions:
     print(f"{p:.2f}")
```

This is a for loop that goes through every predicted value by the code above, then makes the output nicer by formatting it to 2 decimal places.

# **LINEAR REGRESSION (BINARY OUTCOME - NOT IDEAL)**

```python
# ---------- I/O DATA ----------
X = pd.DataFrame({'age': [20, 30, 40, 50, 60, 70]})
y =                      [0, 0, 0, 1, 1, 1]
```

In this model the outcome will be binary, only 0 or 1.

```python
lr = LinearRegression()
lr.fit(X, y)
```

This code will fit the data into a linear regression model, but the Y of the model has only been binary. I wonder if it will only output in 0 and 1 also.

```python
ages_to_predict = pd.DataFrame({'age': [5, 45, 100]})
predictions = lr.predict(ages_to_predict)
```

The model was able to predict values for the new ages; however, it is not very clean, this is because it's still a linear regression model. The model's output was never restricted to 0 and 1.

```python
for p in predictions:
     print(f"{p:.2f}")
```

This is the issue with using a linear regression model for binary outcomes, it still draws a linear relationship between the data points.

# **LOGISTIC REGRESSION (BINARY OUTPUT)**

```python
# --------- IMPORTS ---------
from sklearn.linear_model import LogisticRegression
import numpy as np
```

Logistic regression is a different model that is better for binary outcomes, something that linear regression fails at. It is very good an predicting values between 0 and 1.

```python
X = pd.DataFrame({'age': [20, 30, 40, 50, 60, 70]})
y =                      [0, 0, 0, 1, 1, 1]
```

Even though this is the same data as above, the different model will try to learn where the cutoff is for 1.

```python
lr = LogisticRegression()
lr.fit(X, y)
```

The steps to train the models are the same.

```python
ages_to_predict = pd.DataFrame({'age': [5, 45, 100]})
```

The unknown ages are the same aswell.

```python
probabilities = lr.predict_proba(ages_to_predict)
```

predict_proba gives probabilities for 1 or 0. The first column is the probability of being 0, and the second column is the probability of 1.

```python
for p in probabilities[:, 1]:
     print(f"{p:.2f}")
```

probabilities[:, 1] selects only the probabilities for 1, which are all between 0 and 1. All of this model's output is fit for binary regression.

# **LOGISTIC REGRESSION (NON-LINEAR RELATIONSHIP)**

```python
# --------- IMPORTS ----------
from sklearn.linear_model import LogisticRegression
import pandas as pd
```

This is the same model, but for this example it will instead predict non-linear data, not just a simple cutoff.

```python
X = pd.DataFrame({'age': [20, 30, 40, 50, 60, 70, 80]})
y = [0, 0, 1, 1, 1, 0, 0]
```

Here the data is 1 in the middle but falls back to 0 around 70. Logistic regression may fail at detecting this as it's a complicated pattern.

```python
lr = LogisticRegression()
lr.fit(X, y)
```

We train the model on the non-linear pattern this time.

```python
ages_to_predict = pd.DataFrame({'age': [25, 50, 75]})
probabilities = lr.predict_proba(ages_to_predict)
```

These three new ages sit just before, in the middle, and just after the cutoff points, what will the model do?

```python
for p in probabilities[:, 1]:
     print(f"{p:.2f}")
```

It seems like the probabilites are all the same, because the model is confused by the data's non-linear nature.
